```python
numerical_columns = [
    'district',
    'odds-to-one pre-game | film',
    'kills | books',
    'unintentional or second-hand kills | books',
    'death order | books',
    'rank | books',
    'death order| film',
    'rank | film',
    'winner'
]
```

```python
print("Numerical columns:")
print(numerical_columns)
```

district is stored as a number, but it is really a category (a label for a place, District 7 is not "more" than District 3). Converting it to category makes more sense for grouping.
• winner is stored as 0/1 integers but really means yes/no, so it could be converted to boolean.
• gender only has two values (M/F), so it can be converted to category to save memory and make grouping clearer.
• odds-to-one pre-game | film, kills | books and unintentional or second-hand kills | books are float64 instead of int only because they contain missing values (NaN). They are counts, so the nullable integer type Int64 would describe them better.
• There are no date columns, so no datetime conversion is needed.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
%matplotlib inline
```

```python
by_size = HG_df_clean.groupby("odds-to-one pre-game | film")["rank | books"].median().sort_values()
fig, ax = plt.subplots(figsize=(7, 4))
ax.barh([f"{s} odds" for s in by_size.index], by_size.values, color='gold')
ax.set_title("Odds-to-one Pre-Game (film) by Rank in Books")
ax.set_xlabel("Median Book Rank")
plt.show()
```
