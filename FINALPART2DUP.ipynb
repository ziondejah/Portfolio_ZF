```python
alliance_summary = (
    HG_df.groupby("alliance", dropna=True)
    .agg(
        num_tributes=("name", "count"),
        avg_odds=("odds-to-one pre-game | film", "mean"),
        avg_rank_book=("rank | books", "mean"),
        avg_rank_film=("rank | film", "mean"),
        total_kills=("kills | books", "sum"),
        max_kills=("kills | books", "max")
    )
    .round(2)
)
display(alliance_summary)
```

**what i learned from the 'gender' data:** I decided to choose gender because it showcases how advantages and disadvanteages might have arisen. For an example, if a female ranked higher in the book or film and if that had anything to do with kills. It seems that both M and F got 12 number of tributes between the books and the film. But, the average rank slightly differed, having female contestants at 12.75 and men at 11.33. The average rank and the film actually had females higher at 12.83 and men lower at 11.17. Women got an average of 3 more kills than men but the best rank in the books was actual 2:1 for men.

```python
gender_summary = (
    HG_df.groupby("gender")
    .agg(
        num_tributes=("name", "count"),
        avg_rank_book=("rank | books", "mean"),
        avg_rank_film=("rank | film", "mean"),
        total_kills=("kills | books", "sum"),
        best_rank_book=("rank | books", "min")
    )
    .round(2)
)
display(gender_summary)
```

**what I learned from the 'alliance' data:** The average odds for each alliance for each district varied. The average rank in the film was the highest among district 3 at 16, while distrcit 12 was only 5.5 and district 4 + 10 was 6. This dataset was interesting because it lets us compare the book and film versions of the same Hunger Games and see what is linked to a tribute's survival. One clear pattern is that **alliances matter**: tributes in alliances, and tributes who were willing to kill, generally finished near the top, while most tributes with no alliance died early. The pre-game **betting odds were not a great predictor** – the favorite, Coral (2-to-1), finished 6th in the book, and the winner, Lucy Gray, was a 30-to-1 underdog. Several data issues came up: many missing values in `alliance`, `odds`, and `kills` (mostly because tributes died before the games started), a nearly empty `general comments` column, tied death orders for tributes who died at the same time, and small typos and extra spaces in the text. The dataset is also very small (24 rows), so any patterns should be treated carefully. Questions for Part 2 include: Is there a relationship between betting odds and final rank? How much did the film change the order of deaths, and for which tributes? Do certain districts consistently do better?
